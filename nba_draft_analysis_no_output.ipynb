```python
from balldontlie import BalldontlieAPI
import json
import pandas as pd
import time
from balldontlie.exceptions import RateLimitError
# Initialize the API client
api = BalldontlieAPI(api_key="b0a2f54d-9f0c-4a19-856a-e000cb13668e")

```

```python
all_players = []
cursor = None

while True:
    try:
        resp = api.nba.players.list(per_page=100, cursor=cursor)
        players = resp.data
        all_players.extend([p.model_dump() for p in players])
        print(f"Fetched {len(players)} players – total so far: {len(all_players)}")

        cursor = resp.meta.next_cursor

        if not cursor:
            break

        time.sleep(15)  # Wait to avoid rate limiting
    except RateLimitError:
        print("Hit rate limit – waiting 30 seconds then retrying...")
        time.sleep(30)
```

```python
df_players = pd.DataFrame(all_players)
df_players[['id', 'first_name', 'last_name']].to_csv("all_player_ids.csv", index=False)
print("✅ Saved all player IDs to all_player_ids.csv")
df_players.head()

```
